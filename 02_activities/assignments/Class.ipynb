```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# import the required libraries
```

```python
np.random.seed(613)
x=np.arange(50)
y=np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total_growth_over_time')
ax.set_ylabel('Total_growth')
ax.set_xlabel('Years_since_start')
```

```python
font1 = {'family':'sans-serif','color':'blue','size': 20}
font2 = {'family':'monospace', 'color':'green', 'size':14}
# we are now trying different fonts, sizes and colors
fig,ax =plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total_growth_over_time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total_growth', fontdict = font2)
ax.set_xlabel('Years_since_start', fontdict = font2)
# now tried moving the labels
```

```python
fig, ax = plt.subplots(figsize =(5,3))
ax.plot(x,y,
         marker = '*', 
         markersize = 12,
         color ='indigo',
          linestyle = '--' ,
          linewidth = 2,
          markeredgecolor = '#fa9359',
         markerfacecolor ='#000001')
ax.grid(axis = 'y', color ='green'
        )

#tried out different markers, colors, linestyle. 
# the markeredgecolor and markerfacecolor are newer commands .
#linedge can also be changed as required.
```
