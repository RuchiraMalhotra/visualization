```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file (change the path to where you saved it)
df = pd.read_csv("/Users/rohitmalhotra/Downloads/StrollTO POI Details - 4326.csv")

# Display first few rows and shape to confirm
print(df.shape)
df.head(5)
```

```python
neighbourhoods = df['Neighbourhood'].unique()
print(neighbourhoods)
```

```python
# Choose the Neighbourhoods
neighbourhoods = [
    'West Humber Clairville',
    'Rexdale-Kipling'
]

filtered = df[df['Neighbourhood'].isin(neighbourhoods)]

# Plot each neighbourhood with its own marker and color
plt.figure(figsize=(8,8))

# West Humber Clairville: red stars
whc_tours = filtered[filtered['Neighbourhood'] == 'West Humber Clairville']
plt.scatter(
    [1]*len(whc_tours),
    whc_tours['Title'],  
    marker='*',
    color='red',
    s=80,
    label='West Humber Clairville'
)

# Waterfront Communities-The Island: green stars
wci_tours = filtered[filtered['Neighbourhood'] == 'Rexdale-Kipling']
plt.scatter(
    [2]*len(wci_tours),
    wci_tours['Title'],
    marker='*',
    color='green',
    s=80,
    label='Rexdale-Kipling'
)

# Customize plot
plt.xticks([1,2], ['West Humber Clairville', 'Rexdale-Kipling'], rotation=30, ha='right')
plt.ylabel("Tour Title")
plt.title("StrollTO Tours in Selected Neighbourhoods")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(loc='upper right', bbox_to_anchor=(.65,-0.03))
plt.tight_layout()
plt.show()
```
