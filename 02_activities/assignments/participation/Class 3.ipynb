```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
```

```python
tips = sns.load_dataset("tips")
print (tips)
```

```python
sns.set_style('white')

sns.lineplot(data= tips,
            x='total_bill',
            y='tip')
```

```python
tipgraph = sns.lineplot(data= tips,
            x='total_bill',
            y='tip')

tipgraph.set(title='Tip vs Total Bill',
            xlabel = 'Total Bill($)',
            ylabel = 'Tip amount ($)')

plt.show()
```

```python
fig = plt.subplots(figsize=(10,3))

tipgraph = sns.lineplot(data = tips, 
                    x='total_bill',
                    y='tip',
                    color='hotpink',
                    linestyle='-',
                    linewidth = 1,
                    marker='*',
                    markerfacecolor='indigo')

tipgraph.set(title='Tips vs Total Bill',
            xlabel = 'Total bill($)',
            ylabel = 'Tips($)')

plt.show()
```

```python
tipgraph=sns.scatterplot(data=tips,
                    x='total_bill',
                    y='tip',
                    style='time',
                    hue='day',
                    palette=['red','blue','deepskyblue','yellowgreen'])

tipgraph.set(title='Tips Vs Total Bill',
            xlabel = 'Total Bill($)',
            ylabel = 'Tip amount($)')
```

```python
tipgraph=sns.scatterplot(data=tips,
                    x='total_bill',
                    y='tip',
                    style='day',
                    marker ='*',
                    hue='time',
                    palette=['green','blue'])

tipgraph.set(title='Tips Vs Total Bill',
            xlabel = 'Total Bill($)',
            ylabel = 'Tip amount($)')
```

```python
sns.pairplot(
            data=tips,
            hue='day'
)
```

```python
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter",
    col_wrap=2
)
```

```python
# if we dont use col wrap then it comes as a line
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter"
)
```

```python
import plotly.graph_objects as go
import nbformat

x1=np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1=np.array([110,180,240,99,220])
```

```python
graph =go.Figure()

graph.add_trace(go.Bar(x=x1,y=y1))

graph.update_layout(title = 'Pirate scores',
                    xaxis_title= 'Pirates',
                    yaxis_title = 'Score')

graph.show()
```

```python
graph =go.Figure()

graph.add_trace(go.Scatter(
            x=x1,
            y=y1,
            mode='markers',
            marker = dict(size=15,
                         color='hotpink',
                         opacity =1,
                         line =dict(width=5,color='purple'))
                         ))

graph.update_layout(title= 'Interactive Pirate Plot',
                    xaxis_title= 'Pirates',
                    yaxis_title = 'Score',
                    width = 500,
                    height = 500)
```
